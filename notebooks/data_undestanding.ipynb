```python
## Is 'reserved_tonnes made out of int values?
master['reserves_tonnes_int_true'] = master['reserves_tonnes'].apply(lambda x: x.is_integer())

information_reserves_tones = {
    'int quantity': master['reserves_tonnes_int_true'].sum(),
    'Percentage_of_int': str(master['reserves_tonnes_int_true'].mean() * 100) + '%'
}

print(pd.DataFrame(information_reserves_tones, index=[0]))
```
