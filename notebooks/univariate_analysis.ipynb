# Univariate Analysis

Throughout this notebook, the variables will be observed on their own, in order to understand the beahvior of them

```python
import pandas as pd

master = pd.read_csv('../data/raw/critical_minerals_supply_master.csv')
reference = pd.read_csv('../data/raw/minerals_reference.csv')
```

```python
master.info()
```

## Master dataset:

#### 1. Year:

```python
# Max Year recorded
print('Max year Recorded: ', master['year'].max())

# Min Year Recorded

print('Min year Recorded: ', master['year'].min())

# Quantity of different years
print('Quantity of years recorded: ', master['year'].nunique())
print('Summary: \n', master['year'].value_counts())
```

### 2. Mineral


```python
import matplotlib.pyplot as plt
import seaborn as sns

## Cleaning
master['mineral'] = master['mineral'].str.strip().str.lower()

## Unique values
print('Cantidad de valores unicos: ', master['mineral'].nunique())
print(master['mineral'].unique())


## Quantity of each one
print("="*30)
minerals_count = master['mineral'].value_counts()
print('Quantity of registers per mineral: ', minerals_count)


## top biggest and smallest 3
minerals_big = minerals_count.head(3)
minerals_smll = minerals_count.tail(3)
print("="*30)
print('Biggest:', minerals_big)
print('Smallest:', minerals_smll)



sns.set_style('darkgrid')
sns.countplot(data=master, x='mineral', color='teal')
plt.title('Number of registers per mineral')
plt.xticks(rotation=90)
plt.show()
```

### **Summary** 
Here we can observe that most of the different minerals has a very similar quantity of registers, netherless, some of them highlited by their numbers

#### Top 3 biggest
The analysis of the dataset reveals that the three metals with the highest frequency are copper, nickel, and tin. This observation aligns with the broader context of industrial demand, as these materials

are among the most commercially significant in modern society. Specifically, copper is extensively utilized in electrical applications due to its high conductivity, nickel plays a critical role in the

production of stainless steel, and the combination of copper and tin is fundamental to the creation of bronze. Furthermore, these metals exhibit favorable properties such as ease of melting, shaping, and
recyclability, which contribute to their widespread adoption across various sectors.

### 3. Country

```python
## Liempiza rapida

master['country'] = master['country'].str.strip().str.lower()

print('Cantidad de valores unicos: ', master['country'].nunique())
print(master['country'].unique())


## Cantidad de cada tipo
print("="*30)
country_count = master['country'].value_counts()
print('Number of countrys: ', country_count)


sns.set_style('darkgrid')
order = master['country'].value_counts().index
sns.countplot(data=master, x='country', color='steelblue', order=order)
plt.title('Number of registers per country')
plt.xticks(rotation=90)
plt.show()

```
